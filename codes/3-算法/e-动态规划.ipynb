```python
def counter(decorator):
    def wrapper(*args, **kwargs):
        wrapper.counter += 1
        return decorator(*args, **kwargs)
    wrapper.counter = 0
    return wrapper
```

```python
@counter
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)

fib(20)
print(f'fib(20) 调用了 {fib.counter} 次')
```

```python
memo = {}

@counter
def fib_memo(n):
    if n <= 1:
        return n
    if n not in memo:
        memo[n] = fib_memo(n - 1) + fib_memo(n - 2)
    return memo[n]

fib_memo(20)
print(f'fib_memo(20) 调用了 {fib_memo.counter} 次')
```

```python
memo = {}
def fib_dynamic(n):
    if n <= 1:
        return n
    memo = {0: 0, 1: 1}
    for _ in range(2, n + 1):
        memo[_] = memo[_ - 1] + memo[_ - 2]
    return memo[n]

print(f'fib_dynamic(20) = {fib_dynamic(20)}')
```

```python
mem = {}

def knapsack_dfs_mem(wgt, val, i, c):
    if i < 0 or c == 0:
        return 0

    if (i, c) in mem:
        return mem[(i, c)]

    no = knapsack_dfs_mem(wgt, val, i - 1, c)
    if wgt[i] > c:
        mem[(i, c)] = no
        return no

    yes = val[i] + knapsack_dfs_mem(wgt, val, i - 1, c - wgt[i])
    mem[(i, c)] = max(no, yes)
    return mem[(i, c)]  


wgt = [10, 20, 30, 40, 50]
val = [50, 120, 150, 210, 240]
cap = 50
result = knapsack_dfs_mem(wgt, val, len(wgt) - 1, cap)
print(f"最大价值为：{result}")


```

```python
def knapsack_dp(wgt: list[int], val: list[int], cap: int) -> int:
    n = len(wgt)
    # 初始化dp表
    dp = [[0] * (cap + 1) for _ in range(n + 1)]
    for i in range(1, n + 1):
        for c in range(1, cap + 1):
            if wgt[i - 1] > c:
                # 若超过背包容量，则不选物品 i
                dp[i][c] = dp[i - 1][c]
            else:
                # 不选和选物品 i 这两种方案的较大值
                dp[i][c] = max(dp[i - 1][c], dp[i - 1][c - wgt[i - 1]] + val[i - 1])
    return dp[n][cap]

result = knapsack_dp(wgt, val, cap)
print(f"最大价值为：{result}")
```
