```python
arr = [4, 6, 3, 5, 2, 1]

def bubble_sort(nums):
    n = len(nums)
    for i in range(n - 1, 0, -1):
        for j in range(i):
            if nums[j] > nums[j + 1]:
                nums[j], nums[j + 1] = nums[j + 1], nums[j]

bubble_sort(arr)
print(arr)
```

```python
arr = [4, 6, 3, 5, 2, 1]

def bubble_sort_with_flag(nums):
    n = len(nums)
    for i in range(n - 1, 0, -1):
        flag = False
        
        for j in range(i):
            if nums[j] > nums[j + 1]:
                nums[j], nums[j + 1] = nums[j + 1], nums[j]
                flag = True  
        if not flag:
            break

bubble_sort(arr)
print(arr)
```

```python
arr = [5, 9, 10, 7, 8, 20, 6, 0]

def selection_sort(nums):
    n = len(nums)
    for i in range(n - 1):
        k = i
        for j in range(i + 1, n):
            if nums[j] < nums[k]:
                k = j 
        nums[i], nums[k] = nums[k], nums[i]

selection_sort(arr)
print(arr)
```

```python
arr = [4, 1, 3, 1, 5, 2]

def insert_sort(arr):
    n = len(arr)
    for i in range(1, n):
        for j in range(i, 0, -1):
            if arr[j] < arr[j-1]:
                arr[j], arr[j-1] = arr[j-1], arr[j]
            else:
                break

insert_sort(arr)
print(arr)
```
