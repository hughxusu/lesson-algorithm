```python
def linear_search(arr, target):
    for i in arr:
        if i == target:
            return arr.index(i)
    return -1
```

```python
arr = [5, 1, 3, 1, 5, 2]
print(linear_search(arr, 3))
print(linear_search(arr, 6))
```

```python
def linear_search_2(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
        elif arr[i] > target:
            return -1
    return -1
```

```python
arr = [1, 2, 2, 6, 8, 9]
print(linear_search_2(arr, 3))
print(linear_search_2(arr, 6))
```

```python
def binary_search_recursive(arr, target):
    n = len(arr)
    if n == 0:
        return -1

    mid = n // 2
    if arr[mid] == target:
        return mid
    elif arr[mid] > target:
        return binary_search_recursive(arr[:mid], target)
    else:
        return binary_search_recursive(arr[mid+1:], target)
```

```python
arr = [1, 2, 2, 6, 8, 9]
print(binary_search_recursive(arr, 3))
print(binary_search_recursive(arr, 6))
```

```python
def binary_search(arr, target):
    lo = 0
    hi = len(arr) - 1
    while lo <= hi:
        mid = lo + (hi - lo) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] > target:
            hi = mid - 1
        else:
            lo = mid + 1
    return -1
```

```python
arr = [1, 2, 2, 6, 8, 9]
print(binary_search(arr, 3))
print(binary_search(arr, 6))
```

```python
def max_value(arr):
    max_val = arr[0]
    for i in range(1, len(arr)):
        if arr[i] > max_val:
            max_val = arr[i]
    return max_val
```

```python
def get_list(n):
    return [i for i in range(n)]
```

```python
import time

for i in range(10, 28):
    n = pow(2, i)
    arr = get_list(n)
    start = time.time()
    max_value(arr)
    end = time.time()
    print(f'2^{i}={n}, time={end-start}')
```
