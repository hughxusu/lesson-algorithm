```python
def longest_substring_brute(s):
    n = len(s)
    
    for i in range(n):
        for j in range(i + 1, n + 1):
            substr = s[i:j]
            print(substr, end=" " * 4)
        print()
```

```python

s = "abcabcbb"
print(longest_substring_brute(s))
```
