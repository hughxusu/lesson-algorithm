```python
def bf_search_double_loop(pattern, text):
    if not pattern or not text:
        return -1

    for i in range(len(text) - len(pattern) + 1):
        for j in range(len(pattern)):
            if text[i + j] != pattern[j]:
                break
        else:
            return i
    return -1
```

```python
text = "hello world"
pattern = "world"
print(bf_search_double_loop(pattern, text))
patten = 'python'
print(bf_search_double_loop(patten, text))
```

```python
def bf_search_slice(pattern: str, text: str) -> int:
    if not pattern or not text:
        return -1

    for i in range(len(text) - len(pattern) + 1):
        if text[i:i+len(pattern)] == pattern:
            return i
    return -1
```

```python
text = "hello world"
pattern = "world"
print(bf_search_slice(pattern, text))
patten = 'python'
print(bf_search_slice(patten, text))
```

```python
def bf_search_single_loop(pattern: str, text: str) -> int:
    if not pattern or not text:
        return -1

    ti = 0
    pi = 0

    while ti < len(text) and pi < len(pattern):
        if text[ti] == pattern[pi]:
            ti += 1
            pi += 1
        else:
            ti = ti - pi + 1
            pi = 0

        if pi == len(pattern):
            return ti - pi

    return -1
```

```python
text = "hello world"
pattern = "world"
print(bf_search_single_loop(pattern, text))
patten = 'python'
print(bf_search_single_loop(patten, text))
```

```python
def get_pm_next(pattern):
    m = len(pattern)
    if m == 0: return []
    
    pm_map = [0] * m  
    
    j = 0 
    for i in range(1, m):
        while j > 0 and pattern[i] != pattern[j]:
            j = pm_map[j - 1] 
            
        if pattern[i] == pattern[j]:
            j += 1
            
        pm_map[i] = j

    next_arr = [0] * m
    next_arr[0] = -1
    for i in range(1, m):
        next_arr[i] = pm_map[i - 1]

    return pm_map, next_arr
```

```python
pattern = "ababaf"
print(f"模式串: {pattern}")
pm_map, next_arr = get_pm_next(pattern)
print(f"PM 数组: {pm_map}")
print(f"Next 数组: {next_arr}")
```

```python
def get_next(pattern):
    m = len(pattern)
    next_arr = [0] * m
    next_arr[0] = -1  
    
    i = 0 
    j = -1  
    
    # i = [0, m-1] 
    while i < m - 1:
        if j == -1 or pattern[i] == pattern[j]:
            i += 1
            j += 1
            next_arr[i] = j
        else:
            j = next_arr[j]
            
    return next_arr
```

```python
print(f"模式串: {pattern}")
print(f"Next 数组: {get_next(pattern)}")
```

```python
def kmp_search(pattern: str, text: str) -> int:
    if not pattern or not text:
        return -1

    next_arr = get_next(pattern)
    
    i = 0 
    j = 0 
    
    while i < len(text) and j < len(pattern):
        if j == -1 or text[i] == pattern[j]:
            i += 1
            j += 1
        else:
            j = next_arr[j]
            
    if j == len(pattern):
        return i - j
    else:
        return -1
```

```python
pattern = "ababaf"
text = "acababafdfdasfda"
print(kmp_search(pattern, text))
```
