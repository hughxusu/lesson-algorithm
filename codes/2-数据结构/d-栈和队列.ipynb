```python
class Stack:
    def __init__(self):
        self.__items = []

    def push(self, item):
        self.__items.append(item)

    def pop(self):
        return self.__items.pop()

    def length(self):
        return len(self.__items)
```

```python
stack = Stack()
for i in range(5):
    print(i, end='\t')
    stack.push(i)
```

```python
for i in range(stack.length()):
    print(stack.pop(), end='\t')
```

```python
class Queue:
    def __init__(self):
        self.__head = None
        self.__tail = None
        self.length = 0

    def push(self, value):
        node = {'value': value, 'next': None}
        if self.length == 0:
            self.__head = node  
            self.__tail = node
        else:
            self.__tail['next'] = node
            self.__tail = node
        self.length += 1

    def pop(self):
        if self.length == 0:
            return None
        node = self.__head
        self.__head = node['next']
        self.length -= 1
        return node['value']
```

```python
queue = Queue()
for i in range(5):
    queue.push(i)
    print(i, end='\t')
```

```python
for i in range(queue.length):
    print(queue.pop(), end='\t')
```

```python
class Deque:
    def __init__(self):
        self.__head = None
        self.__tail = None
        self.length = 0

    def push_first(self, value):
        node = {'value': value, 'next': None, 'prev': None}
        if self.length == 0:
            self.__head = node  
            self.__tail = node
        else:
            node['next'] = self.__head
            self.__head['prev'] = node
            self.__head = node
        self.length += 1

    def push_last(self, value):
        node = {'value': value, 'next': None, 'prev': None}
        if self.length == 0:
            self.__head = node  
            self.__tail = node
        else:
            self.__tail['next'] = node
            node['prev'] = self.__tail
            self.__tail = node
        self.length += 1

    def pop_first(self):
        if self.length == 0:
            return None
        node = self.__head
        self.__head = node['next']
        if self.__head:
            self.__head['prev'] = None
        self.length -= 1
        return node['value']

    def pop_last(self):
        if self.length == 0:
            return None
        node = self.__tail
        self.__tail = node['prev']
        if self.__tail:
            self.__tail['next'] = None
        self.length -= 1
        return node['value']
```

```python
deque = Deque()
for i in range(5):
    deque.push_first(i)
    print(i, end='\t')
```

```python
for i in range(5):
    print(deque.pop_last(), end='\t')
```

```python
from collections import deque

deq = deque()

# 添加至队尾
deq.append(2)    
deq.append(4)

# 添加至队首
deq.appendleft(3)  
deq.appendleft(1)

# 弹出队尾元素
print(deq.pop()) 

# 弹出队首元素 
print(deq.popleft())  
```
